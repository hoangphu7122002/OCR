```python
import json
import base64
from PIL import Image
import io
import matplotlib.pyplot as plt

def process_text(metadata):
    content = metadata.get("content", "")
    print("\nText content:")
    print(content)

def process_image(metadata):
    content = metadata.get("content", "")
    if content:
        try:
            # Decode base64 image
            image_data = base64.b64decode(content)
            image = Image.open(io.BytesIO(image_data))
            
            # Hiển thị thông tin ảnh
            print("\nImage Information:")
            print(f"Size: {image.size}")
            print(f"Format: {image.format}")
            
            # Hiển thị ảnh
            plt.figure(figsize=(10, 10))
            plt.imshow(image)
            plt.axis('off')
            plt.show()
        except Exception as e:
            print(f"Error processing image: {e}")

def process_structured(metadata):
    try:
        content_metadata = metadata.get("content_metadata", {})
        table_metadata = metadata.get("table_metadata", {})
        
        print("\nStructured Information:")
        print("-" * 50)
        print("Description:", content_metadata.get("description", ""))
        
        if table_metadata:
            print("\nTable Content:")
            print("-" * 50)
            print(table_metadata.get("table_content", ""))
            
    except Exception as e:
        print(f"Error processing structured data: {e}")

def parse_document_data(json_file_path):
    try:
        with open(json_file_path, 'r', encoding='utf-8') as file:
            data = json.load(file)
    except Exception as e:
        print(f"Error reading JSON file: {e}")
        return

    results = data.get("result", [[]])[0]
    
    for item in results:
        doc_type = item.get("document_type")
        metadata = item.get("metadata", {})
        
        print(f"\n{'='*50}")
        print(f"Document Type: {doc_type}")
        print(f"{'='*50}")
        
        if doc_type == "text":
            process_text(metadata)
        elif doc_type == "image":
            process_image(metadata)
        elif doc_type == "structured":
            process_structured(metadata)
        else:
            print(f"Unknown document type: {doc_type}")
```
